```python
import re
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
for pkg in ['punkt', 'punkt_tab', 'stopwords', 'wordnet', 'averaged_perceptron_tagger',
            'averaged_perceptron_tagger_eng']:
    nltk.download(pkg, quiet=True)
```

```python
text = "The runners were running quickly and the dogs barked loudly. Mr. Smith isn't happy!"
print("Original text:\n", text)
```

```python
# 1. Sentence segmentation
print("\n1. Sentences:", sent_tokenize(text))
```

```python
# 2. Lowercase + remove punctuation
clean = re.sub(r'[^\w\s]', ' ', text.lower())
print("\n2. Lowercased/punctuation removed:", clean)
```

```python
# 3. Tokenization
tokens = word_tokenize(clean)
print("\n3. Tokens:", tokens)
```

```python
# 4. Stop-word removal
stop_words = set(stopwords.words('english'))
filtered = [w for w in tokens if w not in stop_words]
print("\n4. After stop-word removal:", filtered)
```

```python
# 5. Stemming
ps = PorterStemmer()
print("\n5. Stems:", [ps.stem(w) for w in filtered])

# 6.Lemmatization
lem = WordNetLemmatizer()
print("\n6. Lemmas (noun):", [lem.lemmatize(w) for w in filtered])
print("   Lemmas (verb):", [lem.lemmatize(w, pos='v') for w in filtered])
```

```python
# 7. POS tagging
print("\n7. POS tags:", nltk.pos_tag(filtered))
```
